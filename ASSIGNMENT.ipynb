```python
import pandas as pd
import numpy as np
```

```python


df = pd.DataFrame(
    np.random.randint(1, 10, size=(5,3)),
    columns=['A','B','C']
)

df['Product'] = df['A'] * df['B']

print(df)
```

```python

df = pd.DataFrame(
    np.random.randint(1, 20, size=(4,3)),
    columns=['A','B','C']
)

print(df)

print("\nColumn-wise Sum:")
print(df.sum())

print("\nRow-wise Sum:")
print(df.sum(axis=1))
```

```python


df = pd.DataFrame(
    np.random.randint(1, 20, size=(5,3)),
    columns=['A','B','C']
)

df.loc[1,'A'] = np.nan
df.loc[3,'B'] = np.nan

print("Before Filling:")
print(df)

df = df.fillna(df.mean())

print("\nAfter Filling:")
print(df)
```

```python

df = pd.DataFrame(
    np.random.randint(1, 20, size=(6,4)),
    columns=['A','B','C','D']
)

df.loc[1,'A'] = np.nan
df.loc[4,'C'] = np.nan

print("Before Dropping:")
print(df)

df = df.dropna()

print("\nAfter Dropping:")
print(df)
```

```python


df = pd.DataFrame({
    'Category': np.random.choice(['A','B','C'],10),
    'Value': np.random.randint(1,100,10)
})

print(df)

result = df.groupby('Category')['Value'].agg(['sum','mean'])

print("\nAggregated Data:")
print(result)
```

```python


df = pd.DataFrame({
    'Product': np.random.choice(
        ['P1','P2','P3','P4'],10
    ),
    'Category': np.random.choice(
        ['Electronics','Clothing','Food'],10
    ),
    'Sales': np.random.randint(100,1000,10)
})

print(df)

sales = df.groupby('Category')['Sales'].sum()

print("\nTotal Sales:")
print(sales)
```

```python


df1 = pd.DataFrame({
    'ID':[1,2,3],
    'Name':['A','B','C']
})

df2 = pd.DataFrame({
    'ID':[1,2,3],
    'Marks':[80,90,85]
})

merged = pd.merge(df1,df2,on='ID')

print(merged)
```

```python


index = pd.MultiIndex.from_tuples(
    [('A',1),
     ('A',2),
     ('B',1),
     ('B',2)]
)

df = pd.DataFrame(
    np.random.randint(1,100,(4,2)),
    index=index,
    columns=['X','Y']
)

print(df)

print("\nRows of Category A:")
print(df.loc['A'])

print("\nSpecific Row:")
print(df.loc[('B',1)])
```

```python


index = pd.MultiIndex.from_tuples([
('A','A1'),
('A','A2'),
('B','B1'),
('B','B2'),
('C','C1')
],
names=['Category','SubCategory'])

df = pd.DataFrame(
    {'Value':np.random.randint(1,100,5)},
    index=index
)

print(df)

print("\nSum by Category:")
print(df.groupby(level='Category').sum())

print("\nSum by SubCategory:")
print(df.groupby(level='SubCategory').sum())
```

```python


df = pd.DataFrame(
    np.random.randint(1,20,(5,3)),
    columns=['A','B','C']
)

print("Original:")
print(df)

df = df.apply(lambda x: x*2)

print("\nDoubled:")
print(df)
```

```python


df = pd.DataFrame(
    np.random.randint(1,20,(6,3)),
    columns=['A','B','C']
)

df['Total'] = df.apply(
    lambda row:
    row['A'] + row['B'] + row['C'],
    axis=1
)

print(df)
```

```python


s = pd.Series([
    "apple",
    "banana",
    "orange",
    "grapes",
    "mango"
])

print("Original:")
print(s)

print("\nUppercase:")
print(s.str.upper())
```
